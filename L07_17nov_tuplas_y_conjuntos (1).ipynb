# 📚 Parte 1 · Tuplas: datos en posiciones que no cambian

📦 Una tupla es como una caja sellada con casillas numeradas. Se puede **leer** cada casilla, pero no reemplazar su contenido después de crearla.

```python
ficha = ("Lina", "ADSO", 18)
```

- Se escribe con comas y, normalmente, paréntesis. La primera posición es `0`, como en una lista.
- Funcionan `ficha[0]`, `ficha[-1]`, `len(ficha)` y `for dato in ficha`.
- `ficha[0] = "Ana"` da `TypeError`: una tupla es **inmutable**. Tampoco tiene `.append()`.
- La tupla de un solo dato necesita coma: `(5,)`. `(5)` es el número 5.

**Desempacar** significa recibir cada parte en una variable: `nombre, programa, edad = ficha`. Debe haber tantas variables como datos. Una función con `return menor, mayor` devuelve una tupla; por eso se puede escribir `bajo, alto = extremos(lista)`.

````python
ficha = ("Lina", "ADSO", 18)
print(ficha[0])
print(ficha[-1])
nombre, programa, edad = ficha
print(nombre, programa)

def minmax(numeros):
    return min(numeros), max(numeros)

resultado = minmax([5, 2, 9])
print(resultado)
bajo, alto = resultado
print(bajo, alto)
print(type((5,)).__name__)
print(type((5)).__name__)
````

**Qué pasó, línea por línea:** se leyeron la primera y la última casilla; las tres variables recibieron los datos en orden. `minmax` entregó `(2, 9)`, primero como una tupla completa y luego desempacada. La coma distingue `(5,)` de `(5)`. Use una **lista** si necesita cambiar posiciones.

# 📚 Parte 2 · Conjuntos: códigos sin repetidos

🎟️ En una planilla de acceso, un código escrito dos veces cuenta como **una persona**. Un **conjunto** (`set`) guarda cada elemento una sola vez.

```python
registrados = {"A1", "B2", "A1"}  # solo dos códigos distintos
vacio = set()                   # {} sería un diccionario vacío
```

- No hay posiciones: `registrados[0]` da `TypeError`.
- El orden al imprimir un conjunto no está garantizado. `sorted(registrados)` entrega una **lista ordenada**.
- `x in registrados` comprueba pertenencia; `len(registrados)` cuenta elementos diferentes.
- `.add(x)` agrega; `.discard(x)` quita sin fallar si no está. `.remove(x)` sí falla si falta.
- Cadenas, números y tuplas de datos inmutables pueden ser elementos; una lista no.

````python
codigos = {"A1", "B2", "A1"}
print(len(codigos))
print("B2" in codigos)
codigos.add("C3")
codigos.add("A1")
print(sorted(codigos))
codigos.discard("X9")
codigos.discard("B2")
print(sorted(codigos))
print(type(set()).__name__)
````

**Qué pasó:** el segundo `A1` no aumentó el tamaño. `.add('C3')` sí agregó un código; volver a agregar `A1` no cambió nada. `.discard('X9')` no falló. `sorted` permite comparar la salida sin depender del orden interno.

### Ejercicio 3 · Invitados sin duplicados  
**Nivel:** ⭐ Fácil · **Tipo:** ordenar las líneas  
**👥 En pareja:** A pilotea (escribe), B copilotea (dicta y revisa)

Ordene las líneas de la función. Antes, diga en papel qué estructura elimina repetidos y qué función ordena.

**👀 Así debe verse al ejecutarlo:**
```
['Ana', 'Carlos', 'Lina']
```

````python
# Estas líneas están DESORDENADAS. Córtelas y péguelas en el orden correcto (la sangría ya está bien).
def nombres_unicos(nombres):
    distintos = set(nombres)
    return sorted(distintos)

# Pruebas (no las cambie)
print(nombres_unicos(["Lina", "Ana", "Lina", "Carlos", "Ana"]))
````

# 📚 Parte 3 · Comparar dos grupos

Hay inscritos y presentes. Tres operaciones responden preguntas frecuentes:

| Pregunta | Operación | Nombre |
|---|---|---|
| ¿Quién aparece en **alguno** de los grupos? | `inscritos.union(presentes)` | Unión |
| ¿Quién aparece en **ambos**? | `inscritos & presentes` | Intersección |
| ¿Quién está inscrito y **no** vino? | `inscritos - presentes` | Diferencia |

La dirección de la diferencia importa: `presentes - inscritos` son quienes llegaron sin inscribirse. Las operaciones devuelven conjuntos; `sorted(...)` sirve para mostrarlos siempre en el mismo orden.

La unión también se escribe `inscritos | presentes`; ambas formas producen el mismo conjunto.

````python
inscritos = {"A1", "B2", "C3"}
presentes = {"A1", "C3", "D4"}
print(sorted(inscritos | presentes))
print(sorted(inscritos & presentes))
print(sorted(inscritos - presentes))
print(sorted(presentes - inscritos))
````

**Qué pasó:** la unión tiene cuatro códigos; la intersección contiene `A1` y `C3`; faltó `B2`; `D4` vino sin inscripción. Lea cada resta de izquierda a derecha.

# 🧭 ¿Cuál estructura conviene?

| Necesito… | Use | Motivo |
|---|---|---|
| Una fila con posiciones que puedo cambiar | Lista `[]` | Ordenada y modificable |
| Un dato con partes fijas | Tupla `()` | Ordenada e inmutable |
| Códigos sin repetidos o comparar grupos | Conjunto `{...}` / `set()` | Elementos únicos, sin posiciones |

### Ejercicio 7 · El informe que se daña  
**Nivel:** ⭐⭐⭐ Reto · **Tipo:** cazar el error y modificar  
**👥 En pareja:** A pilotea (escribe), B copilotea (dicta y revisa)

**La situación:** el informe debe devolver tres datos, en este orden: inscritos que faltaron, presentes sin inscripción y cantidad de presentes inscritos. El compañero invirtió las diferencias y pidió la posición cero de un conjunto.

1. Dibuje los dos grupos y calcule las tres respuestas en papel.
2. Corrija la función para que devuelva una **tupla de tres datos**. No imprima dentro de la función.
3. Desempaque el resultado y muestre los conjuntos con `sorted()`.

**👀 Así debe verse al ejecutarlo**:
```
['B2']
['D4']
2
```

<details><summary>💡 Ver pista (último recurso: primero intente y relea la explicación)</summary>

Lea cada resta como «los de la izquierda que no están a la derecha». La intersección sirve para contar quienes están en ambos.

</details>

````python
def informe(inscritos, presentes):
    faltaron = presentes - inscritos
    sin_inscripcion = inscritos - presentes
    asistieron = len(inscritos & presentes)
    return faltaron, sin_inscripcion, asistieron

inscritos = {"A1", "B2", "C3"}
presentes = {"A1", "C3", "D4"}
faltaron, sin_inscripcion, asistieron = informe(inscritos, presentes)
print(sorted(faltaron))
print(sorted(sin_inscripcion))
print(asistieron)
````

# 🛠️ Proyecto del día · Control de asistencia de la feria

**Individual. Primero el plan en papel.** Hay códigos inscritos y códigos presentes. Puede llegar alguien sin inscripción. Cada código debe contarse una sola vez.

**Datos iniciales:** `inscritos = {'A1', 'B2', 'C3', 'D4'}` y `presentes = set()`.

| Opción | Qué hace |
|---|---|
| 1. Registrar llegada | Pide código, quita espacios y pasa a mayúsculas. Rechaza vacío y repetido; si es nuevo, lo agrega a presentes. |
| 2. Ver balance | Muestra presentes inscritos, inscritos que faltan y presentes sin inscripción, siempre ordenados. |
| 3. Inscribir código | Pide y limpia código. Rechaza vacío o repetido; si es nuevo, lo agrega a inscritos. |
| 4. Salir | Termina el programa. |

**Funciones obligatorias:** `limpiar(codigo)` devuelve el texto limpio. `balance(inscritos, presentes)` devuelve una **tupla de tres conjuntos** en el orden de la opción 2. El menú llama a esas funciones.

### Plan en papel
1. ¿Qué tengo?, ¿qué piden?, ¿qué herramienta sirve? Explique por qué las dos colecciones son conjuntos.
2. Haga una tabla de las cuatro opciones y qué modifica cada una.
3. Calcule a mano el balance si llegan `A1`, `C3` y `X9`.
4. Escriba la operación que produce cada uno de los tres conjuntos de `balance`.

### Construcción por etapas
1. Escriba y pruebe las funciones sin menú.
2. Añada las opciones 1 y 4. Pruebe un código vacío y uno repetido.
3. Añada la opción 3. Compruebe cómo cambia el balance al inscribir a quien ya llegó.
4. Añada la opción 2. Muestre los tres conjuntos con `sorted()`.
5. Pruebe una opción inválida.

**Prueba final:** registre `A1`, `C3`, `X9` y otra vez `A1`. El balance muestra `['A1', 'C3']`, `['B2', 'D4']`, `['X9']`. Después inscriba `X9`: el tercer grupo queda vacío.

**Rúbrica (5 puntos):** plan en papel 1.0 · validación y repetidos 1.0 · funciones y tupla 1.0 · balance correcto y ordenado 1.5 · prueba y explicación oral 0.5.

````python
# Escriba aquí su código, siguiendo el paso a paso de arriba
inscritos = {'A1', 'B2', 'C3', 'D4'}
presentes = set()

def limpiar(codigo):
    return codigo.strip().upper()

def balance(inscritos, presentes):
    presentes_inscritos = presentes & inscritos
    faltaron = inscritos - presentes
    sin_inscripcion = presentes - inscritos
    return presentes_inscritos, faltaron, sin_inscripcion

while True:
    print("=== CONTROL FERIA ===")
    print("1. Registrar llegada")
    print("2. Ver balance")
    print("3. Inscribir código")
    print("4. Salir")

    opcion = input("Opción: ").strip()

    match opcion:
        case "1":
            codigo = limpiar(input("Código: "))
            if codigo == "":
                print("Código vacío")
            elif codigo in presentes:
                print("Código ya registrado")
            else:
                presentes.add(codigo)
                print(f"Registrado: {codigo}")

        case "2":
            p_inscritos, faltaron, sin_insc = balance(inscritos, presentes)
            print(f"Presentes inscritos: {sorted(p_inscritos)}")
            print(f"Faltaron: {sorted(faltaron)}")
            print(f"Sin inscripción: {sorted(sin_insc)}")

        case "3":
            codigo = limpiar(input("Código: "))
            if codigo == "":
                print("Código vacío")
            elif codigo in inscritos:
                print("Código ya inscrito")
            else:
                inscritos.add(codigo)
                print(f"Inscrito: {codigo}")

        case "4":
            print("Hasta luego.")
            break

        case _:
            print("Opción no válida")
````
